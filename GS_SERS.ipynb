```python
def calcular_economia_energia(consumo_kwh):
    """Calcula o consumo mensal de energia de uma residência de idoso (ex: 150 kWh/mês)
    e compara com a Tarifa Social e o Sistema Solar Fotovoltaico."""

    tarifa_kwh = 0.85   # R$ por kWh

    # Alternativa 1: Tarifa Convencional
    custo_convencional = consumo_kwh * tarifa_kwh

    # Alternativa 2: Tarifa Social (Desconto deaté 65% para baixa renda / BPC)
    desconto_social = 0.65
    custo_social = custo_convencional * (1 - desconto_social)
    economia_social = custo_convencional - custo_social

    # Alternativa 3: Sistema Solar Residencial (Redução de até 80%)
    custo_solar = custo_convencional * 0.20
    economia_solar = custo_convencional - custo_solar

    resposta = f"""
    --- ANÁLISE DE ENERGIA E TARIFA SOCIAL (SERS / SUS) ---

    > **Público Atendido:** Idosos / Famílias de baixa renda.

    **1. Situação Atual (Tarifa Convencional):**
- Consumo informado: {consumo_kwh:.1f} kWh/mês.
- Custo estimado: R$ {custo_convencional:.2f} por mês.

    **2. Alternativa com Tarifa Social (CadÚnico / BPC):**
- Novo custo com desconto de 65%: R$ {custo_social:.2f} por mês.
- **Economia financeira:** R$ {economia_social:.2f} por mês na conta de luz!

    **3. Alternativa com Energia Solar Residencial:**
- Novo custo estimado (taxa mínima): R$ {custo_solar:.2f} por mês.
- **Economia com fonte renovável:** R$ {economia_solar:.2f} por mês.


    > ** Limitações e Condições:**
    > Valores calculados com base na tarifa média de R$ {tarifa_kwh}/kWh. A Tarifa Social exige inscrição ativa no CadÚnico ou recebimento do BPC. A economia solar depende da viabilidade técnica do telhado. Este assistente NÃO é um canal oficial do governo.

    ** Fontes de Referência:**
- Resoluções Normativas da ANEEL / Diretrizes de Eficiência Energética SERS.
- Lei Orgânica da Assistência Social (BPC) e CadÚnico.
    """
    return resposta
```

```python
def assistente_virtual():
    print(" Bem-vindo ao Citizen Assist AI (Módulo Energia & Sustentabilidade)!")
    print("Como posso ajudar você hoje com relação à eficiência energética e tarifas?")
    print("Digite 1 para calcular economia na conta de luz (Tarifa Social / Solar).")
    print("Digite 2 para sair.\n")

    while True:
        opcao = input("Escolha uma opção (1 ou 2): ")

        if opcao == "1":
            try:
                consumo = float(input("Qual é o consumo médio mensal da residência em kWh (ex: 150)? "))
                if consumo < 0:
                    print("Por favor, digite um valor positivo.\n")
                    continue
                resultado = calcular_economia_energia(consumo)
                print(resultado)
                print("\n" + "="*50 + "\n")
            except ValueError:
                print("Por favor, digite um número válido em kWh.\n")
        elif opcao == "2":
            print("Obrigado por usar o Citizen Assist AI! Até logo.")
            break
        else:
            print("Opção inválida. Tente novamente.\n")
```

```python
if __name__ == "__main__":
    assistente_virtual()
```
